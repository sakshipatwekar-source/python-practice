5. with open() =
- Automatically closes the file
- No need to use close() manually.
- Better and safer way to handle files.


syntax:
with open("filename" , "mode") as file:
         operations

```python
with open("sample.txt" , "r") as file:
     print(file.read())
```

"r" = opens the file in read mode
file.read() = reads all the content
with = automatically closes the file after reading.

Checking whether a file is closed or not ?

```python
file = open("sample.txt" , "r")
print(file.closed)
file.close()
print(file.closed)
```

False - the file is open inside the 'with' block.

True - the file is automatically closed after the with block ends.

```python
with open("sample.txt" , "r") as file:
   print(file.closed)

print(file.closed)
```

6. tell() and seek()- These methods help us control the file cursor(the position fromm where python reads or writes).

tell() - tells the current cursor position.

seek() or seek(0) - moves the cursor to a specific position.


```python
with open("sample.txt" , "r") as file:
  print(file.tell())
  print(file.read(5))
  print(file.tell())
  file.seek(0)
  print(file.read(5))
```

In above example :
. file.tell() output is 0 means cursor starts at position  0.

. file.read(5) output I lo means reads first 5 characters.

. file.tell() output 5 means cursor moves to position 5.

. file.seek(0) output 0 means moves cursor to the beginning.

.file.read(5) output I lo means reads the same 5 characters again.


Because above file currently starts with "I love learning python." That's why the output is I lo, not "Hello".


```python
file = open("sample.txt" , "w")
file.write("Hello Python!\nI love coding.\nPython is easy.")
file.close()
```

```python
with open("sample.txt" , "r") as file:
     file.read(5)
     file.seek(0)
     print(file.readline())
```

file.read(5)- reads 5 characters.

file.seek(0)- moves the cursor back to the beginning.

file.readline()- reads the entire first line.

10. Exception handling(try-except):  Sometimes python encounters an error . we use try-except to handle it without stopping the progam.


try: runs the code.  that might cause an error.

except: handles the error if it occurs.

```python
try:
     file = open("sample.txt" , "r")
     print(file.read())
     file.close()

except FileNotFoundError:
   print("File does not exist!")
```
