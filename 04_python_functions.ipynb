2. Function with parameters:

Parameters allow us to pass information into a fuction.

```python
def greet(name):
  print("Hello" , name)

greet("Amit")
greet("Neha")
```

```python
def add(number1 , number2):
  print(number1 + number2)

add(10 , 20)
```

number1 and number2 are parameters.

10 and 20 are arguments.

print(number1 + number2) calculates and displays the sum.

4. Default Parameters = A default parameter has a value that python uses when you don't provide an argument.

```python
def greet(name="Friend"):
    print("Hello" , name)

greet()
greet("Amit")
```

eg. Create a function introduce() with a default parameter city="Pune".

Call it twice:
1. Without passing a city.
2. Passing "Mumbai".

```python
def introduce(city="Pune"):
     print("I live in" , city)

introduce()
introduce("Mumbai")
```

introduce() - We don't provide a city , so python automatically uses "Pune".


introduce("Mumbai") - We provide "Mumbai" , so python uses "Mumbai" insted of "Pune".

5. Multiple parameters

```python
def student(name, age=23):
  print("Name:", name)
  print("Age:", age)

student("Amit")
student("Neha", 25)
```

```python
def student(name, age=23):
  print("Name:", name)
  print("Age:", age)

student("Rahul" , 30)
```

6. Keyword Arguments = We can pass arguments using parameter names.

```python
def student(name, age):
    print(name, age)

student(age=25, name="Priya")
```

```python
def student(name, age):
    print(name, age)

student(age=28 , name="Amit")
```

7. *args = allows a function to accept multiple arguments without knowing how many will be passed.

```python
def add(*numbers):
    print(sum(numbers))

add(10 , 20 , 30)
add(5 , 10 , 15 , 20)
```

```python
def total(*numbers):
    print(sum(numbers))

total(2 , 4 , 6 , 8)
```

*args collects  multiple arguments into a tuple.

8. **kwargs allows a function to accept multiple keyword arguments.

```python
def student(**details):
   print(details)

student(name="Amit" , age=23 , city="Pune")
```

*args = multiple positional arguments(tuple).


**kwargs = multiple keyword arguments(dictionary).

```python
def info(**data):
   print(data["name"])

info(name="Neha" , age=24)
```

Because data["name"] accesses the value associated with the "name" key.

9. Lambda Functions = A lambda is a small , one-line anonymous fuction.

```python
square = lambda x: x * x

print(square(5))
```

```python
cube = lambda x: x * x * x

print(cube(3))
```

```python
def add(a , b):
    return a + b

def subtract(a , b):
   return a - b

def multiply(a , b):
   return a * b

def divide(a , b):
   return a / b

print(add(10 , 5))
print(subtract(10 , 5))
print(multiply(10 , 5))
print(divide(10 , 5))
```
