4. Type Conversion in python

 4.1 Definition : Type conversion means changing a value from one data type to another data type.
  For example:
  .String to Integer
  .Integer to Float
  .Integer to String
  .Integer to Boolean

 4.2 Types of type conversion
    python supports two types of type conversion:
         1. Implicit type conversion
         2. Explicit type conversion

 4.3 Explicit type conversion:
     Manually converting one data type into another using built-in python functions.
     It is also called type casting.

   1.int()- convert to integer
     The int() function converts a suitable value into an integer.
      
      eg. a = "25"
          b = int(a)

          print(b)
          print(type(b))

          output:
          25
          <class 'int'>

   2.float()- convert to float
     The float() function converts a suitable value into a floating-point number.
      
      eg. a = "25"
          b = float(a)

          print(b)
          print(type(b))

          output:
          25.0
          <class 'float'>

   3.str()- convert to string
      The str() function converts a value into a string.
       
       eg. a = 100
           b = str(a)

           print(b)
           print(type(b))

           output:
           100
           <class 'str'>

   4.bool() - convert to boolean
      The bool() function converts a value into either True or False.
        
        eg. a = 1
            b = bool(a)

            print(b)
            print(type(b))

            output:
            True
            <class 'bool'>

 4.4 Implicit type conversion
     It happens automatically when python converts one data type into another without the programmer explicity asking for it.
      
      eg. a = 10
          b = 5.5

          c = a +b
          
          print(c)
          print(type(c))

          output:
          15.5
          <class 'float'>

Use add() to add one item to a set

Remove items from a set = python provides remove() to remove a specific item.

```python
fruits = {"apple" , "banana" , "mango" , "orange"}

fruits.remove("banana")
print(fruits)
```

discard() method = discard() also removes a specific item from a set.

.Iportant difference:
  .remove() gives an error if the item doesn't exist
  
  .discard() does not give an error if the item doesn't exist

set operations: Union (|) combines all unique elements from two sets.

```python
A = {10 , 20 , 30}
B = {30 , 40 , 50}

print(A | B)
```

Set Intersection = Intersection (&) returns only the common elements between two sets

```python
A = {10 , 20 , 30 , 40}
B = {30 , 40 , 50 , 60}

print(A & B)
```

Set Difference = Difference(-) returns elements that are in the first set but not in the second set.

```python
A = {10 , 20 , 30 , 40}
B = {30 , 40 , 50 , 60}

print(A - B)
print(B - A)
```

Symmetric Difference = Symmetric difference(^) returns elements that are in either set , but not in both

```python
A = {10 , 20 , 30 , 40}
B = {30 , 40 , 50 , 60}

print(A ^ B)
```

Check Membership = The 'in' operator checks whether an item exists in a set.

```python
fruits = {"apple" , "banana" , "mango"}

print("banana" in fruits)
print("orange" in fruits)
```

```python
numbers = {10 , 20 , 30 , 40}

print(len(numbers))

numbers.clear()
print(numbers)
```

```python
python_skills = {"variables" , "loops" , "lists" , "tuples"}

sql_skills = {"SQL" , "joins" , "loops" , "lists"}

print(python_skills | sql_skills)
print(python_skills & sql_skills)
print(python_skills - sql_skills)
print(len(python_skills))
```
